```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Lakehouse_Aula") \
    .getOrCreate()

print("Spark iniciado!")
```

```python
dados = [
    (1, "2026-09-01", "Notebook", "Tecnologia", 3500, 2, "SP"),
    (2, "2026-09-01", "Mouse", "Tecnologia", 80, 10, "SP"),
    (3, "2026-09-02", "Teclado", "Tecnologia", 150, 5, "RJ"),
    (4, "2026-09-02", "Cadeira", "Móveis", 900, 3, "MG"),
    (5, "2026-09-03", "Mesa", "Móveis", 1200, 2, "SP"),
    (6, "2026-09-03", "Monitor", "Tecnologia", 1500, 4, "PR"),
    (7, "2026-09-04", "Notebook", "Tecnologia", 3500, 1, "RJ"),
    (8, "2026-09-04", "Cadeira", "Móveis", 900, 2, "SP"),
    (9, "2026-09-05", "Mouse", "Tecnologia", 80, 20, "MG"),
    (10, "2026-09-05", "Mesa", "Móveis", 1200, 1, "RJ")
]

colunas = [
    "id_venda",
    "data",
    "produto",
    "categoria",
    "preco",
    "quantidade",
    "estado"
]

df_bronze = spark.createDataFrame(dados, colunas)

df_bronze.show()
```

```python
df_bronze.count()
```

```python
df_bronze.columns
```

```python
df_bronze.printSchema()
```

```python
df_bronze.select(
    "produto",
    "preco",
    "quantidade"
).show()
```

```python
from pyspark.sql.functions import col

df_silver = df_bronze.withColumn(
    "valor_total",
    col("preco") * col("quantidade")
)

df_silver.show()
```

```python
from pyspark.sql.functions import when

df_silver = df_silver.withColumn(
    "faixa_venda",
    when(col("valor_total") < 1000, "Baixa")
    .when(col("valor_total") <= 5000, "Média")
    .otherwise("Alta")
)

df_silver.show()
```

```python
df_silver.filter(
    col("estado") == "SP"
).show()
```

```python
df_silver.filter(
    col("valor_total") > 3000
).show()
```

```python
df_silver.createOrReplaceTempView("vendas")
```

```python
spark.sql("""
    SELECT *
    FROM vendas
""").show()
```

```python
spark.sql("""
    SELECT produto
    FROM vendas
""").show()
```

```python
spark.sql("""
    SELECT *
    FROM vendas
    WHERE estado = 'SP'
""").show()
```

```python
spark.sql("""
    SELECT *
    FROM vendas
    WHERE valor_total > 3000
""").show()
```

```python
df_gold = spark.sql("""
    SELECT
        categoria,
        SUM(valor_total) AS faturamento
    FROM vendas
    GROUP BY categoria
    ORDER BY faturamento DESC
""")

df_gold.show()
```

```python
df_gold = spark.sql("""
    SELECT
        produto,
        COUNT(*) AS quantidade_vendas,
        SUM(valor_total) AS faturamento
    FROM vendas
    GROUP BY produto
    ORDER BY faturamento DESC
""")

df_gold.show()
```

```python
df_gold_categoria = spark.sql("""
    SELECT
        categoria,
        SUM(valor_total) AS faturamento
    FROM vendas
    GROUP BY categoria
    ORDER BY faturamento DESC
""")

df_gold_categoria.show()
```

```python
df_gold_categoria.toPandas().plot(
    x="categoria",
    y="faturamento",
    kind="bar"
)
```

```python
spark.sql("""
    SELECT
        estado,
        SUM(valor_total) AS faturamento
    FROM vendas
    GROUP BY estado
    ORDER BY faturamento DESC
""").show()

```

```python
spark.sql("""
    SELECT
        categoria,
        SUM(quantidade) AS unidades
    FROM vendas
    GROUP BY categoria
    ORDER BY unidades DESC
""").show()
```
